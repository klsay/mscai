```python
# 1. Repair Excel-corrupted Beat values and standardize strings
beat_map = {"2.00E+01": "2E1", "2.00E+02": "2E2"}
df["Beat"] = df["Beat"].astype(str).str.strip().replace(beat_map)
```

```python
# 2. Convert disguised (0,0) coordinates to NaN for spatial subsets
df.loc[(df["Latitude"] == 0) | (df["Longitude"] == 0), ["Latitude", "Longitude"]] = np.nan
```

```python
# 3. Parse mixed datetime formats and engineer temporal features
df["Start_DT"] = pd.to_datetime(df["Start_Date_Time"], format="mixed", errors="coerce")
df["Dispatch_DT"] = pd.to_datetime(df["Dispatch Date / Time"], format="mixed", errors="coerce")
df["Year"] = df["Start_DT"].dt.year
df["Month"] = df["Start_DT"].dt.month
df["Hour"] = df["Start_DT"].dt.hour
```

```python
# 4. Calculate valid dispatch-to-start response times (excluding negative/extreme noise)
df["Response_Mins"] = (df["Dispatch_DT"] - df["Start_DT"]).dt.total_seconds() / 60.0
df.loc[(df["Response_Mins"] < 0) | (df["Response_Mins"] > 1440), "Response_Mins"] = np.nan
```

```python
# 1. Cast low-cardinality nominal columns to 'category' (saves memory & speeds up EDA)
cat_cols = [
    "Crime Name1", "Crime Name2", "Police District Name",
    "Police District Number", "Agency", "Sector", "Beat", "State"
]
for col in cat_cols:
    df[col] = df[col].astype("category")

# 2. Cast remaining text/identifier columns to explicit 'string' dtype
str_cols = [
    "Offence Code", "NIBRS Code", "Crime Name3", "Block Address",
    "City", "Place", "PRA", "Street Prefix", "Street Name",
    "Street Suffix", "Street Type", "Location"
]
for col in str_cols:
    df[col] = df[col].astype("string")

# 3. Optional: Cast Zip Code & Address Number to nullable integer (Int64) so they don't show '.0'
df["Zip Code"] = df["Zip Code"].astype("Int64")
df["Address Number"] = df["Address Number"].astype("Int64")
```

```python
# 1. Inspect raw unique values and counts
raw_unique = df['Beat'].dropna().unique()
print(f"Raw unique Beat count: {len(raw_unique)}")
print("Sample raw values:", sorted(raw_unique)[:15])

# 2. Clean common formatting anomalies
def clean_beat(val):
    if pd.isna(val):
        return np.nan

    val = str(val).strip().upper()

    # Fix Excel scientific notation corruption for 2E beats
    sci_fixes = {
        '2.00E+01': '2E1', '2E+01': '2E1', '20': '2E1', '20.0': '2E1',
        '2.00E+02': '2E2', '2E+02': '2E2', '200': '2E2', '200.0': '2E2',
        '2.00E+03': '2E3', '2E+03': '2E3', '2000': '2E3', '2000.0': '2E3',
        '2.00E+04': '2E4', '2E+04': '2E4', '20000': '2E4', '20000.0': '2E4',
    }
    if val in sci_fixes:
        return sci_fixes[val]

    # Remove accidental decimal points if read as float (e.g., '1.0' -> '1')
    if val.endswith('.0'):
        val = val[:-2]

    return val

df['Beat_Clean'] = df['Beat'].apply(clean_beat)

# 3. Map Beat prefix (first character) to Official Police District
district_map = {
    '1': '1D - Rockville',
    '2': '2D - Bethesda',
    '3': '3D - Silver Spring',
    '4': '4D - Wheaton',
    '5': '5D - Germantown',
    '6': '6D - Gaithersburg / Montgomery Village',
    '8': '8D - Takoma Park'  # Municipal police department reporting into the dataset
}

df['Mapped_District'] = df['Beat_Clean'].str[0].map(district_map).fillna('Unknown / Unmapped')

# 4. Verify cleaned results
print(f"\nCleaned unique Beat count: {df['Beat_Clean'].nunique()}")
print("\nRecord counts by Mapped District:")
print(df['Mapped_District'].value_counts())

# Optional: View any remaining unusual Beat codes (not matching standard 3-char format like 1A1)
anomalies = df[~df['Beat_Clean'].str.match(r'^[1-8][A-Z]\d$', na=True)]['Beat_Clean'].unique()
print("\nNon-standard Beat codes remaining:", anomalies)
```
