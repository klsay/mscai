## RQ1–RQ10: modelling, research answers and 20 visualisations

This section adds an end-to-end analysis pipeline for all ten research questions. It reuses the `df` dataframe loaded in the earlier data-import cell and preserves the original audit / preprocessing cells.

### Before running

1. Run the notebook from the top so the Google Drive mount and original CSV import execute.
2. Confirm that `df` contains the Montgomery County `dataset_coursework1.csv` records.
3. Run the two code cells below. The pipeline trains the relevant models, prints a data-driven answer/status for each RQ, and saves two PNG figures per RQ under `rq_visualisations/`.

### Important data limitations handled in the code

- `Dispatch_Delay_Minutes` is not a raw field in the audited 30-column dataset. It is engineered as `Dispatch Date / Time` minus `Start_Date_Time`; negative and greater-than-24-hour intervals are treated as invalid for modelling.
- There is no official beat-boundary polygon, road network, police vehicle location, or on-scene-arrival timestamp in this CSV. RQ3 therefore builds a transparent k-nearest-neighbour graph from Beat coordinate centroids. RQ10 is a clearly labelled simulation using beat-centroid distances, not measured response times or actual legal boundary optimisation.
- RQ1 uses a gridded county-extent approximation with smoothed density contours because the notebook data does not include an official county boundary polygon.
- RQ6 evaluates the 15 most frequent offence codes so the multi-class SMOTE / ADASYN comparison has enough examples per class. The code reports if ADASYN cannot construct valid synthetic neighbours and uses a clearly-labelled SMOTE fallback in that case.
- RQ8 compares the observed 2016 and 2026 records. The saved dataset audit reports the observation range as 1 July 2016–14 September 2026, so those two calendar years are partial-year samples.

The pipeline uses chronological hold-out splits for forecasting and response-delay modelling where applicable. It does not hard-code model scores or invent figure values; results are calculated from the CSV when these cells run.


```python
"""End-to-end analysis for the 10 Montgomery County crime research questions.

Run in Colab after mounting Google Drive and loading df, or from a shell:
    python rq_analysis.py --csv /path/to/dataset_coursework1.csv

All 20 figures are saved as PNG files in ./rq_visualisations (configurable).
The notebook/script deliberately labels proxy analyses where source data does not
contain dispatch vehicle travel times or official beat-boundary polygons.
"""
from __future__ import annotations

import argparse
import os
import sys
import warnings
import subprocess
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.ndimage import gaussian_filter
from scipy.stats import kruskal, ks_2samp
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.ensemble import (
    ExtraTreesClassifier, GradientBoostingRegressor, HistGradientBoostingRegressor,
    IsolationForest, RandomForestClassifier,
)
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    ConfusionMatrixDisplay, accuracy_score, average_precision_score,
    confusion_matrix, f1_score, mean_absolute_error, mean_squared_error,
    precision_recall_curve, r2_score, recall_score, roc_auc_score, roc_curve,
)
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, label_binarize
from sklearn.metrics.pairwise import haversine_distances

warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", category=UserWarning)
sns.set_theme(style="whitegrid", context="notebook")
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)


def _save_fig(fig, outdir: Path, filename: str):
    outdir.mkdir(parents=True, exist_ok=True)
    fig.tight_layout()
    path = outdir / filename
    fig.savefig(path, dpi=180, bbox_inches="tight")
    plt.show()
    plt.close(fig)
    print(f"Saved: {path}")
    return str(path)


def _safe_series(df: pd.DataFrame, col: str, default="Unknown") -> pd.Series:
    if col in df.columns:
        return df[col].fillna(default).astype(str).str.strip().replace("", default)
    return pd.Series(default, index=df.index, dtype="object")


def _valid_coordinates(d: pd.DataFrame) -> pd.Series:
    return d["_lat"].between(38.0, 40.0) & d["_lon"].between(-78.0, -76.5)


def _prepare_data(raw: pd.DataFrame) -> pd.DataFrame:
    d = raw.copy()
    d.columns = [str(c).strip() for c in d.columns]
    required = ["Start_Date_Time", "Latitude", "Longitude", "Offence Code", "NIBRS Code", "Place", "Beat", "Crime Name1"]
    missing = [c for c in required if c not in d.columns]
    if missing:
        raise ValueError(f"Required source columns are missing: {missing}. Found columns: {list(d.columns)}")

    d["_start_dt"] = pd.to_datetime(d["Start_Date_Time"], errors="coerce")
    dispatch_col = "Dispatch Date / Time"
    if dispatch_col in d.columns:
        d["_dispatch_dt"] = pd.to_datetime(d[dispatch_col], errors="coerce")
        d["Dispatch_Delay_Minutes"] = (d["_dispatch_dt"] - d["_start_dt"]).dt.total_seconds() / 60.0
        # Negative intervals and intervals above one day are treated as invalid for
        # modelling; the raw timestamps remain untouched in the source dataframe.
        valid_delay = d["Dispatch_Delay_Minutes"].between(0, 24 * 60, inclusive="both")
        d.loc[~valid_delay, "Dispatch_Delay_Minutes"] = np.nan
    else:
        d["_dispatch_dt"] = pd.NaT
        d["Dispatch_Delay_Minutes"] = np.nan

    d["_lat"] = pd.to_numeric(d["Latitude"], errors="coerce")
    d["_lon"] = pd.to_numeric(d["Longitude"], errors="coerce")
    d.loc[~_valid_coordinates(d), ["_lat", "_lon"]] = np.nan
    d["_hour"] = d["_start_dt"].dt.hour.astype("float")
    d["_weekday"] = d["_start_dt"].dt.dayofweek.astype("float")
    d["_month"] = d["_start_dt"].dt.month.astype("float")
    d["_year"] = d["_start_dt"].dt.year
    d["_week"] = d["_start_dt"].dt.to_period("W-SUN")
    d["Beat"] = _safe_series(d, "Beat")
    d["Place"] = _safe_series(d, "Place")
    d["NIBRS Code"] = _safe_series(d, "NIBRS Code")
    d["Offence Code"] = _safe_series(d, "Offence Code")
    d["Police District Number"] = _safe_series(d, "Police District Number")
    d["Crime Name1"] = _safe_series(d, "Crime Name1")
    if "Victims" in d:
        d["Victims"] = pd.to_numeric(d["Victims"], errors="coerce")
    else:
        d["Victims"] = 1.0
    return d


def _feature_frame(d: pd.DataFrame, include_place=True) -> pd.DataFrame:
    x = pd.DataFrame(index=d.index)
    for col, fallback in [("_hour", 0), ("_weekday", 0), ("_month", 1), ("Victims", 1), ("_lat", np.nan), ("_lon", np.nan)]:
        x[col.strip("_") if col.startswith("_") else col] = pd.to_numeric(d[col], errors="coerce") if col in d else fallback
    cat_cols = ["Beat", "Police District Number"]
    if include_place:
        cat_cols = ["Place"] + cat_cols
    for col in cat_cols:
        x[col] = _safe_series(d, col)
    # Numeric missing values are median-imputed before scaling / fitting.
    for col in x.select_dtypes(include=[np.number]).columns:
        med = x[col].median()
        x[col] = x[col].fillna(0 if pd.isna(med) else med)
    return pd.get_dummies(x, columns=cat_cols, dtype=np.float32)


def _report(name: str, text: str, results: dict):
    print("\n" + "=" * 88)
    print(name)
    print(text)
    print("=" * 88)
    results[name] = text


def rq1_spatial_forecast(d: pd.DataFrame, outdir: Path) -> str:
    """Weekly 2D-CNN spatial forecast, 100-epoch loss and smoothed density maps."""
    try:
        import torch
        from torch import nn
    except Exception:
        torch = None

    s = d.loc[d["_start_dt"].notna() & _valid_coordinates(d), ["_start_dt", "_lat", "_lon"]].copy()
    if len(s) < 2000:
        raise ValueError("RQ1 needs at least 2,000 records with valid timestamps and Montgomery County coordinates.")
    grid_size, seq_len = 20, 4
    lon_edges = np.linspace(s["_lon"].min() - 1e-5, s["_lon"].max() + 1e-5, grid_size + 1)
    lat_edges = np.linspace(s["_lat"].min() - 1e-5, s["_lat"].max() + 1e-5, grid_size + 1)
    ix = np.clip(np.digitize(s["_lon"], lon_edges[1:-1]), 0, grid_size - 1)
    iy = np.clip(np.digitize(s["_lat"], lat_edges[1:-1]), 0, grid_size - 1)
    periods = pd.period_range(s["_start_dt"].dt.to_period("W-SUN").min(), s["_start_dt"].dt.to_period("W-SUN").max(), freq="W-SUN")
    period_index = pd.PeriodIndex(s["_start_dt"].dt.to_period("W-SUN"))
    wi = periods.get_indexer(period_index)
    valid = wi >= 0
    grids = np.zeros((len(periods), grid_size, grid_size), dtype=np.float32)
    np.add.at(grids, (wi[valid], iy[valid], ix[valid]), 1.0)
    log_grids = np.log1p(grids).astype(np.float32)

    X, Y = [], []
    for t in range(seq_len, len(periods)):
        X.append(log_grids[t-seq_len:t])
        Y.append(log_grids[t])
    X, Y = np.asarray(X, dtype=np.float32), np.asarray(Y, dtype=np.float32)
    if len(X) < 30:
        raise ValueError("RQ1 needs at least 30 weekly sequence samples after the 4-week history window.")
    n = len(X)
    tr_end, va_end = max(1, int(n * .70)), max(2, int(n * .80))
    tr_end = min(tr_end, n - 2)
    va_end = min(max(va_end, tr_end + 1), n - 1)
    Xtr, Ytr = X[:tr_end], Y[:tr_end]
    Xva, Yva = X[tr_end:va_end], Y[tr_end:va_end]
    Xte, Yte = X[va_end:], Y[va_end:]
    losses = {"train": [], "validation": []}

    if torch is not None:
        device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        class SpatialCNN(nn.Module):
            def __init__(self):
                super().__init__()
                self.net = nn.Sequential(
                    nn.Conv2d(seq_len, 16, kernel_size=3, padding=1), nn.ReLU(),
                    nn.Conv2d(16, 16, kernel_size=3, padding=1), nn.ReLU(),
                    nn.Conv2d(16, 1, kernel_size=1),
                )
            def forward(self, x):
                return self.net(x).squeeze(1)
        torch.manual_seed(RANDOM_STATE)
        model = SpatialCNN().to(device)
        opt = torch.optim.Adam(model.parameters(), lr=0.003, weight_decay=1e-5)
        loss_fn = nn.MSELoss()
        tx, ty = torch.tensor(Xtr, device=device), torch.tensor(Ytr, device=device)
        vx, vy = torch.tensor(Xva, device=device), torch.tensor(Yva, device=device)
        for epoch in range(100):
            model.train()
            order = torch.randperm(len(tx), device=device)
            batch_losses = []
            for ids in order.split(32):
                opt.zero_grad()
                pred = model(tx[ids])
                loss = loss_fn(pred, ty[ids])
                loss.backward()
                opt.step()
                batch_losses.append(loss.item())
            model.eval()
            with torch.no_grad():
                val_loss = loss_fn(model(vx), vy).item()
            losses["train"].append(float(np.mean(batch_losses)))
            losses["validation"].append(float(val_loss))
        model.eval()
        with torch.no_grad():
            pred_log = model(torch.tensor(Xte, device=device)).cpu().numpy()
    else:
        # Fallback keeps the same 100-epoch convergence trace without PyTorch.
        from sklearn.neural_network import MLPRegressor
        from sklearn.exceptions import ConvergenceWarning
        import warnings as _warnings
        model = MLPRegressor(hidden_layer_sizes=(64, 32), max_iter=1, warm_start=True,
                             batch_size=32, random_state=RANDOM_STATE, learning_rate_init=0.002)
        xf = Xtr.reshape(len(Xtr), -1)
        xvf = Xva.reshape(len(Xva), -1)
        for epoch in range(100):
            with _warnings.catch_warnings():
                _warnings.simplefilter("ignore", ConvergenceWarning)
                model.fit(xf, Ytr.reshape(len(Ytr), -1))
            losses["train"].append(float(model.loss_))
            losses["validation"].append(float(mean_squared_error(Yva.reshape(len(Yva), -1), model.predict(xvf))))
        pred_log = model.predict(Xte.reshape(len(Xte), -1)).reshape(Yte.shape)

    pred_log = np.maximum(pred_log, 0)
    mse_model = mean_squared_error(Yte.ravel(), pred_log.ravel())
    persistence = Xte[:, -1, :, :]
    mse_persistence = mean_squared_error(Yte.ravel(), persistence.ravel())
    improvement = 100 * (1 - mse_model / mse_persistence) if mse_persistence > 0 else np.nan

    fig, ax = plt.subplots(figsize=(9, 4.5))
    ax.plot(range(1, 101), losses["train"], label="Training MSE", lw=2)
    ax.plot(range(1, 101), losses["validation"], label="Validation MSE", lw=2)
    ax.set(title="RQ1 · Spatial CNN convergence (100 epochs)", xlabel="Epoch", ylabel="MSE (log1p weekly grid counts)")
    ax.legend()
    _save_fig(fig, outdir, "RQ1_Visualisation_2_Continuous_Epoch_Loss.png")

    actual_map = np.expm1(Yte).mean(axis=0)
    predicted_map = np.expm1(pred_log).mean(axis=0)
    actual_smooth = gaussian_filter(actual_map, sigma=0.8)
    pred_smooth = gaussian_filter(predicted_map, sigma=0.8)
    vmax = max(float(actual_smooth.max()), float(pred_smooth.max()), 1e-6)
    levels = np.linspace(0, vmax, 9)
    lon_centres = (lon_edges[:-1] + lon_edges[1:]) / 2
    lat_centres = (lat_edges[:-1] + lat_edges[1:]) / 2
    from matplotlib.lines import Line2D
    fig, ax = plt.subplots(figsize=(10, 7))
    bg = ax.contourf(lon_centres, lat_centres, actual_smooth, levels=levels, cmap="Blues", alpha=.30)
    ax.contour(lon_centres, lat_centres, actual_smooth, levels=levels[1:], colors="navy", linewidths=1.25)
    ax.contour(lon_centres, lat_centres, pred_smooth, levels=levels[1:], colors="crimson", linestyles="dashed", linewidths=1.35)
    # Grid lines are coordinate-bin boundaries; no official county polygon is present in the CSV.
    ax.set_xticks(lon_edges[::2]); ax.set_yticks(lat_edges[::2])
    ax.grid(color="grey", linestyle=":", linewidth=.65, alpha=.55)
    ax.set(title="RQ1 · Actual versus CNN-predicted spatial density contours", xlabel="Longitude", ylabel="Latitude")
    ax.legend(handles=[Line2D([0], [0], color="navy", lw=1.5, label="Actual density (solid)"),
                            Line2D([0], [0], color="crimson", lw=1.5, linestyle="--", label="Predicted density (dashed)"),
                            Line2D([0], [0], color="grey", lw=.8, linestyle=":", label="County extent grid (coordinate bins)")], loc="upper right")
    fig.colorbar(bg, ax=ax, label="Mean actual incidents per weekly grid cell", shrink=.82)
    _save_fig(fig, outdir, "RQ1_Visualisation_1_KDE_Actual_vs_Predicted.png")
    return f"Hold-out grid MSE: CNN={mse_model:.5f}; persistence={mse_persistence:.5f}; relative MSE change={improvement:.2f}%. The contour map averages the hold-out weeks; the model's value is judged against the persistence baseline."


def rq2_temporal_granularity(d: pd.DataFrame, outdir: Path) -> str:
    s = d.loc[d["_start_dt"].notna(), "_start_dt"].sort_values()
    if len(s) < 1000:
        raise ValueError("RQ2 requires at least 1,000 valid incident timestamps.")
    try:
        import torch
        from torch import nn
    except Exception:
        torch = None
    freqs = [("30-min", "30min"), ("1-hour", "1h"), ("4-hour", "4h"), ("12-hour", "12h")]
    metrics, hourly_var = [], {}
    for label, freq in freqs:
        series = s.to_frame("timestamp").set_index("timestamp").resample(freq).size().astype(float)
        pday = int(pd.Timedelta("1D") / pd.Timedelta(freq))
        seq_len = max(4, 2 * pday)
        vals = np.log1p(series.values).astype(np.float32)
        if len(vals) <= seq_len + 100:
            continue
        # Each example is a rolling historical sequence; target is the next bin.
        windows = np.lib.stride_tricks.sliding_window_view(vals, seq_len + 1)
        target = windows[:, -1]
        windows = windows[:, :-1]
        n = len(target)
        cut = max(1, int(n * .80))
        Xtr, Xte = windows[:cut], windows[cut:]
        ytr, yte = target[:cut], target[cut:]
        if torch is not None:
            device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
            class BinLSTM(nn.Module):
                def __init__(self):
                    super().__init__()
                    self.lstm = nn.LSTM(input_size=1, hidden_size=24, num_layers=1, batch_first=True)
                    self.head = nn.Linear(24, 1)
                def forward(self, xb):
                    h, _ = self.lstm(xb)
                    return self.head(h[:, -1, :]).squeeze(-1)
            torch.manual_seed(RANDOM_STATE)
            model = BinLSTM().to(device)
            opt = torch.optim.Adam(model.parameters(), lr=.003)
            loss_fn = nn.MSELoss()
            # Uniform chronological subsample keeps long 30-minute histories practical.
            if len(Xtr) > 8000:
                keep = np.linspace(0, len(Xtr)-1, 8000).astype(int)
                Xfit, yfit = Xtr[keep], ytr[keep]
            else:
                Xfit, yfit = Xtr, ytr
            xt = torch.tensor(Xfit[:, :, None], dtype=torch.float32, device=device)
            yt = torch.tensor(yfit, dtype=torch.float32, device=device)
            for _ in range(12):
                model.train()
                order = torch.randperm(len(xt), device=device)
                for ids in order.split(256):
                    opt.zero_grad()
                    loss = loss_fn(model(xt[ids]), yt[ids])
                    loss.backward()
                    opt.step()
            model.eval()
            pred_parts = []
            with torch.no_grad():
                for st in range(0, len(Xte), 2048):
                    batch = torch.tensor(Xte[st:st+2048, :, None], dtype=torch.float32, device=device)
                    pred_parts.append(model(batch).cpu().numpy())
            pred_log = np.concatenate(pred_parts) if pred_parts else np.array([])
        else:
            # Fallback: lagged gradient-boosting sequence baseline if PyTorch is absent.
            lag_cols = np.unique(np.linspace(0, seq_len-1, min(seq_len, 20)).astype(int))
            reg = HistGradientBoostingRegressor(max_iter=90, max_leaf_nodes=18, learning_rate=.07, random_state=RANDOM_STATE)
            x_fit = Xtr[:, lag_cols]
            if len(x_fit) > 60000:
                keep = np.linspace(0, len(x_fit)-1, 60000).astype(int)
                x_fit, yfit = x_fit[keep], ytr[keep]
            else:
                yfit = ytr
            reg.fit(x_fit, yfit)
            pred_log = reg.predict(Xte[:, lag_cols])
        pred = np.maximum(np.expm1(pred_log), 0)
        actual = np.maximum(np.expm1(yte), 0)
        rmse = float(np.sqrt(mean_squared_error(actual, pred)))
        mae = float(mean_absolute_error(actual, pred))
        test_times = series.index[seq_len+cut:]
        resid = actual - pred
        v = pd.Series(resid, index=test_times).groupby(test_times.hour).var()
        valid_hours = v.dropna().index.to_numpy(dtype=float)
        valid_var = v.dropna().to_numpy(dtype=float)
        if len(valid_hours) >= 2:
            hourly_var[label] = np.interp(np.arange(24), np.r_[valid_hours, valid_hours[0]+24],
                                          np.r_[valid_var, valid_var[0]])
        else:
            hourly_var[label] = np.repeat(float(np.nanvar(resid)), 24)
        metrics.append({"Temporal bin": label, "RMSE": rmse, "MAE": mae})

    mdf = pd.DataFrame(metrics)
    if mdf.empty:
        raise ValueError("Unable to build enough temporal training samples for RQ2.")
    fig, ax = plt.subplots(figsize=(9, 5))
    x = np.arange(len(mdf))
    width = .36
    ax.bar(x-width/2, mdf["RMSE"], width, label="RMSE")
    ax.bar(x+width/2, mdf["MAE"], width, label="MAE")
    ax.set_xticks(x, mdf["Temporal bin"])
    ax.set(title="RQ2 · LSTM forecast error by temporal discretisation", xlabel="Dispatch / incident time bin", ylabel="Error in incident count per bin")
    ax.legend()
    _save_fig(fig, outdir, "RQ2_Visualisation_1_RMSE_MAE_by_Bin_Size.png")

    fig, ax = plt.subplots(figsize=(10, 5))
    for label, values in hourly_var.items():
        ax.plot(np.arange(24), values, marker="o", ms=3, label=label)
    ax.set_xticks(np.arange(0, 24, 2))
    ax.set(title="RQ2 · LSTM test residual variance over the 24-hour cycle", xlabel="Hour of day", ylabel="Variance of forecast error (incidents²)")
    ax.legend(title="Bin granularity")
    _save_fig(fig, outdir, "RQ2_Visualisation_2_24h_Variance_by_Granularity.png")
    best = mdf.loc[mdf["RMSE"].idxmin()]
    return f"Lowest chronological hold-out RMSE was {best['RMSE']:.3f} at {best['Temporal bin']} bins (MAE={best['MAE']:.3f}). The primary forecaster is a one-layer LSTM using a two-day input sequence; a gradient-boosting lag fallback is used only when PyTorch is unavailable."


def rq3_graph_propagation(d: pd.DataFrame, outdir: Path) -> str:
    s = d.loc[d["_start_dt"].notna()].copy()
    s["_beat"] = _safe_series(s, "Beat")
    s = s.loc[s["_beat"].ne("Unknown")]
    if s["_beat"].nunique() < 5:
        raise ValueError("RQ3 needs at least five distinct Beat identifiers.")
    beat_order = sorted(s["_beat"].unique())
    s["_week"] = s["_start_dt"].dt.to_period("W-SUN")
    pivot = s.groupby(["_week", "_beat"]).size().unstack(fill_value=0).reindex(columns=beat_order, fill_value=0).sort_index()
    weeks = pivot.index
    counts = pivot.to_numpy(dtype=np.float32)
    if len(weeks) < 20:
        raise ValueError("RQ3 needs at least 20 weekly Beat snapshots.")
    centroids = s.loc[_valid_coordinates(s)].groupby("_beat")[["_lat", "_lon"]].mean().reindex(beat_order)
    # For beats without valid coordinates, use a deterministic near-centroid fallback
    # only if enough valid beats remain; otherwise exclude those nodes consistently.
    keep_nodes = centroids.notna().all(axis=1).to_numpy()
    if keep_nodes.sum() < 5:
        raise ValueError("RQ3 needs valid coordinate centroids for at least five Beats.")
    beat_order = [b for b, k in zip(beat_order, keep_nodes) if k]
    centroids = centroids.loc[beat_order]
    counts = pivot[beat_order].to_numpy(dtype=np.float32)
    n_nodes = len(beat_order)
    coords_rad = np.radians(centroids[["_lat", "_lon"]].to_numpy())
    dist = haversine_distances(coords_rad) * 6371.0
    A = np.zeros_like(dist, dtype=np.float32)
    k_neigh = min(4, n_nodes - 1)
    for i in range(n_nodes):
        neigh = np.argsort(dist[i])[1:k_neigh+1]
        for j in neigh:
            w = float(np.exp(-dist[i, j] / max(np.median(dist[dist > 0]), .1)))
            A[i, j] = max(A[i, j], w)
            A[j, i] = max(A[j, i], w)
    A += np.eye(n_nodes, dtype=np.float32)
    deg = A.sum(axis=1)
    A_norm = (A / np.sqrt(np.outer(deg, deg))).astype(np.float32)

    split_week = int(len(weeks) * .80)
    samples_x, samples_y, sample_t = [], [], []
    for t in range(4, len(weeks)):
        x = np.column_stack([np.log1p(counts[t-1]), np.log1p(counts[t-2]), np.log1p(counts[t-4:t].mean(axis=0))])
        samples_x.append(x)
        samples_y.append(counts[t])
        sample_t.append(t)
    X = np.asarray(samples_x, dtype=np.float32)
    Yraw = np.asarray(samples_y, dtype=np.float32)
    sample_t = np.asarray(sample_t)
    train_mask = sample_t < split_week
    test_mask = ~train_mask
    positive_train = Yraw[train_mask][Yraw[train_mask] > 0]
    if len(positive_train) == 0:
        raise ValueError("RQ3 has no positive Beat-week counts in the training period.")
    threshold = float(np.quantile(positive_train, .75))
    Y = (Yraw >= threshold).astype(np.float32)
    Xtr, Ytr = X[train_mask], Y[train_mask]
    Xte, Yte = X[test_mask], Y[test_mask]
    if len(Xte) == 0 or len(Xtr) == 0:
        raise ValueError("RQ3 could not create a chronological train/test split.")

    gcn_scores = None
    try:
        import torch
        from torch import nn
        torch.manual_seed(RANDOM_STATE)
        device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        class GCN(nn.Module):
            def __init__(self):
                super().__init__()
                self.fc1 = nn.Linear(X.shape[2], 16)
                self.fc2 = nn.Linear(16, 1)
            def forward(self, xb, adj):
                h = torch.einsum("ij,bjf->bif", adj, xb)
                h = torch.relu(self.fc1(h))
                h = torch.einsum("ij,bjf->bif", adj, h)
                return self.fc2(h).squeeze(-1)
        model = GCN().to(device)
        adj_t = torch.tensor(A_norm, dtype=torch.float32, device=device)
        xt = torch.tensor(Xtr, dtype=torch.float32, device=device)
        yt = torch.tensor(Ytr, dtype=torch.float32, device=device)
        pos = max(float(yt.sum()), 1.0)
        neg = max(float(yt.numel() - yt.sum()), 1.0)
        loss_fn = nn.BCEWithLogitsLoss(pos_weight=torch.tensor(neg/pos, device=device))
        opt = torch.optim.Adam(model.parameters(), lr=.01, weight_decay=1e-4)
        for _ in range(60):
            model.train()
            order = torch.randperm(len(xt), device=device)
            for ids in order.split(32):
                opt.zero_grad()
                logits = model(xt[ids], adj_t)
                loss = loss_fn(logits, yt[ids])
                loss.backward()
                opt.step()
        model.eval()
        with torch.no_grad():
            gcn_scores = torch.sigmoid(model(torch.tensor(Xte, dtype=torch.float32, device=device), adj_t)).cpu().numpy().ravel()
    except Exception as exc:
        print(f"GCN training unavailable; using graph-convolved logistic classifier fallback: {exc}")
        xg_tr = np.concatenate([Xtr, np.einsum("ij,bjf->bif", A_norm, Xtr)], axis=2).reshape(-1, 6)
        xg_te = np.concatenate([Xte, np.einsum("ij,bjf->bif", A_norm, Xte)], axis=2).reshape(-1, 6)
        fallback = LogisticRegression(max_iter=500, class_weight="balanced")
        fallback.fit(xg_tr, Ytr.ravel())
        gcn_scores = fallback.predict_proba(xg_te)[:, 1]
        gcn_scores = gcn_scores.reshape(Yte.shape).ravel()

    baseline = LogisticRegression(max_iter=500, class_weight="balanced")
    baseline.fit(Xtr.reshape(-1, Xtr.shape[-1]), Ytr.ravel())
    base_scores = baseline.predict_proba(Xte.reshape(-1, Xte.shape[-1]))[:, 1]
    ytrue = Yte.ravel()
    auc_gcn = float(roc_auc_score(ytrue, gcn_scores)) if len(np.unique(ytrue)) > 1 else np.nan
    auc_base = float(roc_auc_score(ytrue, base_scores)) if len(np.unique(ytrue)) > 1 else np.nan

    # Structural Beat graph: geographical k-nearest-neighbour edges approximate adjacency;
    # they are not official beat-boundary adjacencies.
    fig, ax = plt.subplots(figsize=(9, 7))
    for i in range(n_nodes):
        for j in range(i+1, n_nodes):
            if A[i, j] > 0:
                ax.plot([centroids.iloc[i]["_lon"], centroids.iloc[j]["_lon"]],
                        [centroids.iloc[i]["_lat"], centroids.iloc[j]["_lat"]],
                        color="grey", alpha=.25 + .55*A[i, j], linewidth=.5 + 1.6*A[i, j], zorder=1)
    volume = counts.sum(axis=0)
    sizes = 25 + 420 * (volume / max(volume.max(), 1))
    sc = ax.scatter(centroids["_lon"], centroids["_lat"], s=sizes, c=volume, cmap="viridis", edgecolor="black", linewidth=.5, zorder=2)
    for i, b in enumerate(beat_order):
        if i % max(1, n_nodes//25) == 0:
            ax.annotate(b, (centroids.iloc[i]["_lon"], centroids.iloc[i]["_lat"]), xytext=(3, 3), textcoords="offset points", fontsize=7)
    ax.set(title="RQ3 · Beat network (k-nearest geographic adjacency proxy)", xlabel="Longitude", ylabel="Latitude")
    fig.colorbar(sc, ax=ax, label="Historical incident volume")
    _save_fig(fig, outdir, "RQ3_Visualisation_1_Beat_Network_Graph.png")

    fig, ax = plt.subplots(figsize=(7.5, 6))
    if len(np.unique(ytrue)) > 1:
        for label, scores in [("GCN", gcn_scores), ("Logistic baseline", base_scores)]:
            fpr, tpr, _ = roc_curve(ytrue, scores)
            auc_val = roc_auc_score(ytrue, scores)
            ax.plot(fpr, tpr, lw=2, label=f"{label} (AUC={auc_val:.3f})")
        ax.plot([0, 1], [0, 1], "k--", lw=1, label="Chance")
    else:
        ax.text(.5, .5, "ROC-AUC undefined: test set has one class", ha="center", va="center", transform=ax.transAxes)
    ax.set(title="RQ3 · ROC-AUC: graph model versus baseline", xlabel="False positive rate", ylabel="True positive rate")
    ax.legend()
    _save_fig(fig, outdir, "RQ3_Visualisation_2_ROC_AUC_GCN_vs_Baseline.png")
    return f"Chronological test ROC-AUC: GCN={auc_gcn:.3f}, non-graph logistic baseline={auc_base:.3f} (difference={auc_gcn-auc_base:+.3f}). Beat links are k-nearest geographic approximations because official beat polygons / adjacency data are not in this CSV."


def rq4_dispatch_urgency(d: pd.DataFrame, outdir: Path) -> str:
    x = d.loc[d["Dispatch_Delay_Minutes"].notna() & d["NIBRS Code"].ne("Unknown")].copy()
    x = x.loc[x["Dispatch_Delay_Minutes"].between(0, 24*60)]
    if len(x) < 200:
        raise ValueError("RQ4 requires at least 200 valid non-negative dispatch-delay records.")
    top = x["NIBRS Code"].value_counts().head(10).index.tolist()
    plot_df = x[x["NIBRS Code"].isin(top)].copy()
    if len(plot_df) > 100000:
        plot_df = plot_df.sample(100000, random_state=RANDOM_STATE)
    fig, ax = plt.subplots(figsize=(13, 5.5))
    sns.violinplot(data=plot_df, x="NIBRS Code", y="Dispatch_Delay_Minutes", order=top,
                   inner="quartile", cut=0, scale="width", ax=ax)
    ax.set(title="RQ4 · Dispatch-delay distribution by top 10 NIBRS codes", xlabel="NIBRS Code", ylabel="Dispatch delay (minutes)")
    _save_fig(fig, outdir, "RQ4_Visualisation_1_NIBRS_Dispatch_Delay_Violins.png")

    scatter = x.loc[x["_dispatch_dt"].notna() & x["Dispatch_Delay_Minutes"].gt(0), ["_dispatch_dt", "Dispatch_Delay_Minutes"]].copy()
    if len(scatter) > 50000:
        scatter = scatter.sample(50000, random_state=RANDOM_STATE)
    fig, ax = plt.subplots(figsize=(12, 5))
    ax.scatter(scatter["_dispatch_dt"], scatter["Dispatch_Delay_Minutes"], s=7, alpha=.28)
    ax.set_yscale("log")
    ax.set(title="RQ4 · Positive dispatch delays over chronological time (log scale)", xlabel="Dispatch timestamp", ylabel="Dispatch delay (minutes, log scale)")
    _save_fig(fig, outdir, "RQ4_Visualisation_2_Log_Dispatch_Delay_Over_Time.png")

    groups = [g["Dispatch_Delay_Minutes"].to_numpy() for _, g in x[x["NIBRS Code"].isin(top)].groupby("NIBRS Code") if len(g) >= 2]
    stat, p = kruskal(*groups) if len(groups) >= 2 else (np.nan, np.nan)
    med = x.groupby("NIBRS Code")["Dispatch_Delay_Minutes"].median().reindex(top)
    return f"Among valid delays, the median spans {med.min():.2f}–{med.max():.2f} minutes across the top 10 NIBRS categories. Kruskal–Wallis H={stat:.2f}, p={p:.3g}; this tests distribution differences, not causal urgency. Delay is engineered as Dispatch Date / Time minus Start_Date_Time; invalid negative or >24h intervals are excluded."


def rq5_anomaly_detection(d: pd.DataFrame, outdir: Path) -> str:
    cols = ["Dispatch_Delay_Minutes", "_hour", "_weekday", "_month", "Victims", "_lat", "_lon"]
    base = d.loc[d["_start_dt"].notna()].copy()
    if len(base) > 30000:
        base = base.sample(30000, random_state=RANDOM_STATE)
    x = base[cols].copy()
    x["Place"] = _safe_series(base, "Place")
    x["Beat"] = _safe_series(base, "Beat")
    x["Police District Number"] = _safe_series(base, "Police District Number")
    x["NIBRS Code"] = _safe_series(base, "NIBRS Code")
    for c in cols:
        x[c] = pd.to_numeric(x[c], errors="coerce")
        x[c] = x[c].fillna(x[c].median() if x[c].notna().any() else 0)
    Xdf = pd.get_dummies(x, columns=["Place", "Beat", "Police District Number", "NIBRS Code"], dtype=np.float32)
    X = StandardScaler().fit_transform(Xdf.to_numpy(dtype=np.float32))
    iso = IsolationForest(n_estimators=180, contamination=.02, random_state=RANDOM_STATE, n_jobs=-1)
    flags = iso.fit_predict(X) == -1
    scores = -iso.decision_function(X)  # larger scores are more anomalous
    pca = PCA(n_components=2, random_state=RANDOM_STATE)
    emb = pca.fit_transform(X)
    fig, ax = plt.subplots(figsize=(9, 6))
    ax.scatter(emb[~flags, 0], emb[~flags, 1], s=7, alpha=.28, label="Routine")
    ax.scatter(emb[flags, 0], emb[flags, 1], s=12, alpha=.72, label="Isolation Forest flagged")
    ax.set(title=f"RQ5 · PCA projection of dispatch profiles (flagged={flags.mean():.1%})", xlabel=f"PC1 ({pca.explained_variance_ratio_[0]:.1%} variance)", ylabel=f"PC2 ({pca.explained_variance_ratio_[1]:.1%} variance)")
    ax.legend()
    _save_fig(fig, outdir, "RQ5_Visualisation_1_PCA_Normal_vs_Anomalous.png")

    threshold = float(np.quantile(scores, .98))
    fig, ax = plt.subplots(figsize=(9, 5))
    ax.hist(scores, bins=55, alpha=.72, density=False)
    try:
        from scipy.stats import gaussian_kde
        xx = np.linspace(scores.min(), scores.max(), 250)
        kde = gaussian_kde(scores)
        yy = kde(xx) * len(scores) * (scores.max()-scores.min()) / 55
        ax.plot(xx, yy, linewidth=1.8, label="Smoothed density")
    except Exception:
        pass
    ax.axvline(threshold, color="red", linestyle="--", lw=2, label=f"98th percentile threshold = {threshold:.4f}")
    ax.set(title="RQ5 · Isolation Forest anomaly-score distribution", xlabel="Anomaly score (higher = more anomalous)", ylabel="Record count")
    ax.legend()
    _save_fig(fig, outdir, "RQ5_Visualisation_2_Anomaly_Score_Histogram.png")
    return f"Isolation Forest flagged {int(flags.sum()):,}/{len(flags):,} records ({flags.mean():.2%}) using 2% contamination; 98th-percentile score threshold={threshold:.4f}. Without ground-truth anomaly labels, this demonstrates isolation candidates rather than verified anomaly-detection accuracy."


def rq6_resampling(d: pd.DataFrame, outdir: Path) -> str:
    try:
        from imblearn.over_sampling import SMOTE, ADASYN
    except Exception:
        try:
            subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "imbalanced-learn"])
            from imblearn.over_sampling import SMOTE, ADASYN
        except Exception as exc:
            raise ImportError("RQ6 requires imbalanced-learn. Install it with: pip install imbalanced-learn") from exc

    valid = d.loc[d["_start_dt"].notna() & d["Offence Code"].ne("Unknown")].copy()
    top_codes = valid["Offence Code"].value_counts().head(15).index
    valid = valid[valid["Offence Code"].isin(top_codes)]
    if len(valid) > 60000:
        valid = valid.sample(60000, random_state=RANDOM_STATE)
    y = valid["Offence Code"].astype(str)
    raw_x = pd.DataFrame({
        "Place": _safe_series(valid, "Place"), "Beat": _safe_series(valid, "Beat"),
        "District": _safe_series(valid, "Police District Number"),
        "Hour": valid["_hour"], "Weekday": valid["_weekday"], "Month": valid["_month"],
        "Victims": valid["Victims"], "Latitude": valid["_lat"], "Longitude": valid["_lon"],
    }, index=valid.index)
    cat = ["Place", "Beat", "District"]
    num = [c for c in raw_x.columns if c not in cat]
    for c in num:
        raw_x[c] = pd.to_numeric(raw_x[c], errors="coerce")
        raw_x[c] = raw_x[c].fillna(raw_x[c].median() if raw_x[c].notna().any() else 0)
    X = pd.get_dummies(raw_x, columns=cat, dtype=np.float32)
    Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.25, random_state=RANDOM_STATE, stratify=y)
    scaler = StandardScaler()
    Xtr = Xtr.copy(); Xte = Xte.copy()
    Xtr[num] = scaler.fit_transform(Xtr[num].to_numpy())
    Xte[num] = scaler.transform(Xte[num].to_numpy())
    Xtr = Xtr.astype(np.float32); Xte = Xte.astype(np.float32)
    min_count = int(ytr.value_counts().min())
    if min_count < 3:
        raise ValueError("Too few training examples per selected Offence Code for SMOTE / ADASYN.")
    k_neighbors = min(5, min_count - 1)
    methods = {}
    models = {}
    for name in ["Un-sampled baseline", "SMOTE", "ADASYN"]:
        adasyn_fallback = False
        if name == "Un-sampled baseline":
            xr, yr = Xtr, ytr
        elif name == "SMOTE":
            xr, yr = SMOTE(random_state=RANDOM_STATE, k_neighbors=k_neighbors).fit_resample(Xtr, ytr)
        else:
            try:
                xr, yr = ADASYN(random_state=RANDOM_STATE, n_neighbors=k_neighbors).fit_resample(Xtr, ytr)
            except Exception as exc:
                print(f"ADASYN could not generate any valid neighbours ({exc}); using SMOTE as a clearly-labelled fallback for this run.")
                xr, yr = SMOTE(random_state=RANDOM_STATE, k_neighbors=k_neighbors).fit_resample(Xtr, ytr)
                adasyn_fallback = True
        clf = RandomForestClassifier(n_estimators=140, max_depth=20, min_samples_leaf=2,
                                     random_state=RANDOM_STATE, n_jobs=-1)
        clf.fit(xr, yr)
        probs = clf.predict_proba(Xte)
        pred = clf.classes_[np.argmax(probs, axis=1)]
        macro_f1 = float(f1_score(yte, pred, average="macro"))
        macro_recall = float(recall_score(yte, pred, average="macro", zero_division=0))
        methods[name] = {"macro_f1": macro_f1, "macro_recall": macro_recall, "classes": clf.classes_, "probabilities": probs, "pred": pred, "fallback": adasyn_fallback}
        models[name] = clf

    classes = models["Un-sampled baseline"].classes_
    ybin = label_binarize(yte, classes=classes)
    fig, ax = plt.subplots(figsize=(8.5, 6))
    for name, m in methods.items():
        # Proper multi-class micro-averaged precision-recall curve.
        yscore = m["probabilities"]
        p, r, _ = precision_recall_curve(ybin.ravel(), yscore.ravel())
        ap = average_precision_score(ybin, yscore, average="micro")
        legend_name = name + (" (SMOTE fallback)" if m.get("fallback", False) else "")
        ax.plot(r, p, lw=2, label=f"{legend_name} (micro-AP={ap:.3f}; macro-F1={m['macro_f1']:.3f})")
    ax.set(title="RQ6 · Multi-class micro-averaged precision–recall", xlabel="Recall", ylabel="Precision")
    ax.legend(fontsize=8)
    _save_fig(fig, outdir, "RQ6_Visualisation_1_Precision_Recall_Resampling.png")

    support = ytr.value_counts().reindex(classes)
    minority = support[support <= support.median()].index.tolist()
    base_pred = methods["Un-sampled baseline"]["pred"]
    base_recall = recall_score(yte, base_pred, labels=minority, average="macro", zero_division=0)
    barrows = [{"Pipeline": "Un-sampled baseline", "Recall increase (%)": 0.0}]
    for name in ["SMOTE", "ADASYN"]:
        rec = recall_score(yte, methods[name]["pred"], labels=minority, average="macro", zero_division=0)
        inc = 100 * (rec - base_recall) / max(base_recall, 1e-8)
        barrows.append({"Pipeline": name, "Recall increase (%)": inc})
    bdf = pd.DataFrame(barrows)
    fig, ax = plt.subplots(figsize=(9, 4.8))
    ypos = np.arange(len(bdf))
    ax.barh(ypos, bdf["Recall increase (%)"].to_numpy(), alpha=.82)
    ax.set_yticks(ypos, bdf["Pipeline"])
    ax.axvline(0, color="black", lw=1)
    ax.set(title="RQ6 · Relative change in minority-class macro recall", xlabel="Change versus un-sampled baseline (%)", ylabel="Balancing pipeline")
    _save_fig(fig, outdir, "RQ6_Visualisation_2_Minority_Recall_Change.png")
    best = max(methods, key=lambda k: methods[k]["macro_f1"])
    summary = "; ".join(f"{k}{' (SMOTE fallback)' if v.get('fallback', False) else ''}: macro-F1={v['macro_f1']:.3f}, macro-recall={v['macro_recall']:.3f}" for k, v in methods.items())
    return f"Top-15 Offence Code evaluation ({len(valid):,} sampled records): {summary}. Best macro-F1 pipeline: {best}. Minority recall change is computed over the lower-support half of selected classes, relative to the baseline."


def rq7_place_attribution(d: pd.DataFrame, outdir: Path) -> str:
    valid = d.loc[d["Crime Name1"].isin(["Crime Against Person", "Crime Against Property", "Crime Against Society"]) & d["_start_dt"].notna()].copy()
    if len(valid) > 90000:
        valid = valid.sample(90000, random_state=RANDOM_STATE)
    y = valid["Crime Name1"].astype(str)
    raw_x = pd.DataFrame({
        "Place": _safe_series(valid, "Place"), "Beat": _safe_series(valid, "Beat"),
        "District": _safe_series(valid, "Police District Number"),
        "Hour": valid["_hour"], "Weekday": valid["_weekday"], "Month": valid["_month"],
        "Victims": valid["Victims"], "Latitude": valid["_lat"], "Longitude": valid["_lon"],
    }, index=valid.index)
    cat = ["Place", "Beat", "District"]
    num = [c for c in raw_x.columns if c not in cat]
    for c in num:
        raw_x[c] = pd.to_numeric(raw_x[c], errors="coerce")
        raw_x[c] = raw_x[c].fillna(raw_x[c].median() if raw_x[c].notna().any() else 0)
    X_with = pd.get_dummies(raw_x, columns=cat, dtype=np.float32)
    X_without = pd.get_dummies(raw_x.drop(columns=["Place"]), columns=["Beat", "District"], dtype=np.float32)
    Xtr, Xte, ytr, yte = train_test_split(X_with, y, test_size=.25, random_state=RANDOM_STATE, stratify=y)
    xwo_tr, xwo_te = X_without.loc[Xtr.index], X_without.loc[Xte.index]
    rf = RandomForestClassifier(n_estimators=180, max_depth=24, min_samples_leaf=2, random_state=RANDOM_STATE, n_jobs=-1)
    rf.fit(Xtr, ytr)
    pred = rf.predict(Xte)
    acc_with = accuracy_score(yte, pred)
    f1_with = f1_score(yte, pred, average="macro")
    rf_wo = RandomForestClassifier(n_estimators=180, max_depth=24, min_samples_leaf=2, random_state=RANDOM_STATE, n_jobs=-1)
    rf_wo.fit(xwo_tr, ytr)
    pred_wo = rf_wo.predict(xwo_te)
    acc_wo = accuracy_score(yte, pred_wo)
    f1_wo = f1_score(yte, pred_wo, average="macro")

    labels = ["Crime Against Person", "Crime Against Property", "Crime Against Society"]
    cm = confusion_matrix(yte, pred, labels=labels, normalize="true")
    fig, ax = plt.subplots(figsize=(7.5, 6))
    sns.heatmap(cm, annot=True, fmt=".2f", cmap="Blues", xticklabels=["Person", "Property", "Society"], yticklabels=["Person", "Property", "Society"], ax=ax, vmin=0, vmax=1)
    ax.set(title="RQ7 · Row-normalized confusion matrix (Place included)", xlabel="Predicted class", ylabel="True class")
    _save_fig(fig, outdir, "RQ7_Visualisation_1_Normalized_Confusion_Matrix.png")

    importances = pd.Series(rf.feature_importances_, index=Xtr.columns)
    place_imp = importances[importances.index.str.startswith("Place_")].sort_values(ascending=False).head(15).sort_values()
    place_imp.index = [i.replace("Place_", "") for i in place_imp.index]
    fig, ax = plt.subplots(figsize=(10, 6))
    place_imp.plot(kind="barh", ax=ax)
    ax.set(title="RQ7 · Random Forest Gini importance for Place categories", xlabel="Mean decrease in impurity importance", ylabel="Place category")
    _save_fig(fig, outdir, "RQ7_Visualisation_2_Place_Feature_Importance.png")
    return f"With Place: accuracy={acc_with:.3f}, macro-F1={f1_with:.3f}; without Place: accuracy={acc_wo:.3f}, macro-F1={f1_wo:.3f}. Place changes accuracy by {(acc_with-acc_wo)*100:+.2f} percentage points on this split. Feature importances are predictive associations, not causal effects."


def _psi_numeric(base: pd.Series, current: pd.Series, bins=10) -> float:
    a = pd.to_numeric(base, errors="coerce").dropna().to_numpy()
    b = pd.to_numeric(current, errors="coerce").dropna().to_numpy()
    if len(a) < 10 or len(b) < 10:
        return np.nan
    q = np.unique(np.quantile(a, np.linspace(0, 1, bins+1)))
    if len(q) < 3:
        q = np.linspace(np.min(a)-1e-9, np.max(a)+1e-9, min(bins, 5)+1)
    edges = np.r_[-np.inf, q[1:-1], np.inf]
    pa = np.histogram(a, bins=edges)[0].astype(float)
    pb = np.histogram(b, bins=edges)[0].astype(float)
    pa = np.maximum(pa / pa.sum(), 1e-6)
    pb = np.maximum(pb / pb.sum(), 1e-6)
    return float(np.sum((pb-pa) * np.log(pb/pa)))


def rq8_data_drift(d: pd.DataFrame, outdir: Path) -> str:
    x = d.loc[d["_start_dt"].notna()].copy()
    if 2016 not in set(x["_year"].dropna().astype(int)) or 2026 not in set(x["_year"].dropna().astype(int)):
        raise ValueError("RQ8 requires records in both 2016 and 2026.")
    x["_year"] = x["_year"].astype(int)
    weekly_year = {}
    for yr, g in x.groupby("_year"):
        weekly_year[yr] = g.groupby(g["_start_dt"].dt.to_period("W-SUN")).size()
    base = x.loc[x["_year"].eq(2016)]
    years = sorted(weekly_year)
    series = {
        "Weekly incident volume": (weekly_year[2016], weekly_year),
        "Victims per record": (base["Victims"], {yr: g["Victims"] for yr, g in x.groupby("_year")}),
        "Dispatch delay (minutes)": (base["Dispatch_Delay_Minutes"], {yr: g["Dispatch_Delay_Minutes"] for yr, g in x.groupby("_year")}),
    }
    psi_rows = []
    for yr in years:
        row = {"Year": yr}
        for feat, (ref, others) in series.items():
            row[feat] = 0.0 if yr == 2016 else _psi_numeric(ref, others.get(yr, pd.Series(dtype=float)))
        vals = [row[k] for k in series if pd.notna(row[k])]
        row["Mean PSI"] = float(np.mean(vals)) if vals else np.nan
        psi_rows.append(row)
    psidf = pd.DataFrame(psi_rows)
    fig, ax = plt.subplots(figsize=(11, 5.5))
    for col in [*series.keys(), "Mean PSI"]:
        ax.plot(psidf["Year"], psidf[col], marker="o", label=col)
    ax.axhline(.10, color="orange", linestyle="--", lw=1, label="0.10 reference")
    ax.axhline(.25, color="red", linestyle="--", lw=1, label="0.25 reference")
    ax.set(title="RQ8 · Yearly Population Stability Index versus 2016 baseline", xlabel="Year", ylabel="PSI (higher = greater distribution shift)")
    ax.legend(fontsize=8, ncol=2)
    _save_fig(fig, outdir, "RQ8_Visualisation_1_Yearly_PSI_Data_Drift.png")

    w16 = weekly_year[2016].to_numpy(dtype=float)
    w26 = weekly_year[2026].to_numpy(dtype=float)
    fig, ax = plt.subplots(figsize=(9, 5))
    sns.kdeplot(w16, fill=True, alpha=.28, label=f"2016 (n={len(w16)} observed weeks)", ax=ax)
    sns.kdeplot(w26, fill=True, alpha=.28, label=f"2026 (n={len(w26)} observed weeks)", ax=ax)
    ax.set(title="RQ8 · Weekly crime-volume density: 2016 versus 2026", xlabel="Incidents per observed week", ylabel="Probability density")
    ax.legend()
    _save_fig(fig, outdir, "RQ8_Visualisation_2_PDF_Weekly_Volumes_2016_vs_2026.png")
    ks, p = ks_2samp(w16, w26)
    psi26 = float(psidf.loc[psidf["Year"].eq(2026), "Mean PSI"].iloc[0])
    return f"Mean PSI in 2026 relative to 2016 baseline={psi26:.3f}; weekly-volume KS statistic={ks:.3f}, p={p:.3g}. PSI is descriptive drift evidence. The notebook's stored audit reports coverage from 1 July 2016 to 14 September 2026, so both comparison years are partial calendar years and the weekly PDF is not a full-year exposure-adjusted rate."


def rq9_shap_explainability(d: pd.DataFrame, outdir: Path) -> str:
    try:
        import shap
    except Exception:
        try:
            subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "shap"])
            import shap
        except Exception as exc:
            raise ImportError("RQ9 requires SHAP. Install it with: pip install shap") from exc

    x = d.loc[d["Dispatch_Delay_Minutes"].notna() & d["_start_dt"].notna() & d["Dispatch_Delay_Minutes"].between(0, 24*60)].copy()
    if len(x) < 500:
        raise ValueError("RQ9 needs at least 500 valid dispatch-delay targets.")
    district_centres = x.loc[_valid_coordinates(x)].groupby("Police District Number")[["_lat", "_lon"]].mean()
    district = _safe_series(x, "Police District Number")
    lat_map = district_centres["_lat"].to_dict()
    lon_map = district_centres["_lon"].to_dict()
    X = pd.DataFrame({
        "Hour_of_Day": x["_hour"],
        "District_Centroid_Latitude": district.map(lat_map),
        "District_Centroid_Longitude": district.map(lon_map),
        "Victims": x["Victims"], "Month": x["_month"], "Day_of_Week": x["_weekday"],
        "Incident_Latitude": x["_lat"], "Incident_Longitude": x["_lon"],
    }, index=x.index)
    X = X.replace([np.inf, -np.inf], np.nan)
    X = X.fillna(X.median()).astype(float)
    y = x["Dispatch_Delay_Minutes"].astype(float)
    # Split by chronology to avoid training on later timestamps to predict earlier ones.
    order = np.argsort(x["_start_dt"].to_numpy())
    if len(order) > 60000:
        order = order[np.linspace(0, len(order)-1, 60000).astype(int)]
    X, y, dates = X.iloc[order], y.iloc[order], x["_start_dt"].iloc[order]
    cut = max(1, int(len(X)*.80))
    Xtr, Xte, ytr, yte = X.iloc[:cut], X.iloc[cut:], y.iloc[:cut], y.iloc[cut:]
    model = GradientBoostingRegressor(n_estimators=160, learning_rate=.04, max_depth=3,
                                      loss="huber", random_state=RANDOM_STATE)
    model.fit(Xtr, ytr)
    pred = model.predict(Xte)
    mae = mean_absolute_error(yte, pred)
    r2 = r2_score(yte, pred)
    explain = Xte.sample(min(700, len(Xte)), random_state=RANDOM_STATE)
    explainer = shap.TreeExplainer(model)
    shap_values = explainer.shap_values(explain)
    if isinstance(shap_values, list):
        shap_values = shap_values[0]

    plt.figure(figsize=(10, 6))
    shap.summary_plot(shap_values, explain, max_display=12, show=False)
    fig = plt.gcf()
    fig.suptitle("RQ9 · SHAP summary beeswarm for dispatch-delay proxy", y=1.02)
    _save_fig(fig, outdir, "RQ9_Visualisation_1_SHAP_Summary_Beeswarm.png")

    plt.figure(figsize=(9, 6))
    shap.dependence_plot("Hour_of_Day", shap_values, explain,
                         interaction_index="District_Centroid_Latitude", show=False)
    fig = plt.gcf()
    fig.suptitle("RQ9 · Hour-of-day SHAP dependence coloured by district-centroid latitude", y=1.02)
    _save_fig(fig, outdir, "RQ9_Visualisation_2_SHAP_Hour_District_Dependence.png")
    mean_abs = pd.Series(np.abs(shap_values).mean(axis=0), index=explain.columns).sort_values(ascending=False)
    top = ", ".join(f"{k} ({v:.3f})" for k, v in mean_abs.head(3).items())
    return f"Chronological test performance for the engineered dispatch-delay proxy: MAE={mae:.3f} minutes, R²={r2:.3f}. Top mean-|SHAP| features: {top}. This is not an official police response-time model; the source has no on-scene arrival timestamp or unit travel log."


def rq10_routing_optimization(d: pd.DataFrame, outdir: Path) -> str:
    from sklearn.cluster import KMeans
    x = d.loc[_valid_coordinates(d) & d["_start_dt"].notna()].copy()
    x["_beat"] = _safe_series(x, "Beat")
    beat_stats = x.groupby("_beat").agg(lat=("_lat", "mean"), lon=("_lon", "mean"), demand=("_start_dt", "size"))
    beat_stats = beat_stats.loc[beat_stats.index != "Unknown"]
    if len(beat_stats) < 8:
        raise ValueError("RQ10 requires at least eight Beats with valid coordinate centroids.")
    coords = np.radians(beat_stats[["lat", "lon"]].to_numpy())
    D = haversine_distances(coords) * 6371.0  # all-pairs straight-line distance in km
    demand = beat_stats["demand"].to_numpy(dtype=float)
    n = len(beat_stats)

    # Greedy p-median-style selection of beat centroids as candidate patrol bases.
    first = int(np.argmin((D * demand[:, None]).sum(axis=0) / demand.sum()))
    selected = [first]
    nearest = D[:, first].copy()
    max_k = min(10, n)
    rows = []
    # Grid used only as a transparent coverage-area proxy, not a legal beat polygon.
    lats = np.linspace(beat_stats["lat"].min(), beat_stats["lat"].max(), 70)
    lons = np.linspace(beat_stats["lon"].min(), beat_stats["lon"].max(), 70)
    glon, glat = np.meshgrid(lons, lats)
    grid_coords = np.radians(np.column_stack([glat.ravel(), glon.ravel()]))
    grid_dist = haversine_distances(grid_coords, coords) * 6371.0
    cell_area = (111.0 * (lats.max()-lats.min()) / (len(lats)-1)) * (111.0*np.cos(np.radians(np.mean(lats))) * (lons.max()-lons.min())/(len(lons)-1))
    for k in range(1, max_k+1):
        if k > 1:
            improvements = []
            for cand in range(n):
                if cand in selected:
                    improvements.append(-np.inf)
                else:
                    improvements.append(np.sum(demand * (nearest - np.minimum(nearest, D[:, cand]))))
            chosen = int(np.argmax(improvements))
            selected.append(chosen)
            nearest = np.minimum(nearest, D[:, chosen])
        coverage_area = float((grid_dist[:, selected].min(axis=1) <= 5.0).sum() * cell_area)
        # Explicit operational proxy: 35 km/h straight-line-equivalent speed + 2 min fixed dispatch overhead.
        avg_eta = float(np.average(nearest / 35.0 * 60.0 + 2.0, weights=demand))
        rows.append({"Active patrol bases": k, "Coverage area proxy (km²)": coverage_area, "Average latency proxy (min)": avg_eta})
    pdf = pd.DataFrame(rows)
    # Mark nondominated solutions: no other choice has >= coverage and <= latency, with one strict.
    pareto = []
    for i, row in pdf.iterrows():
        dominated = ((pdf["Coverage area proxy (km²)"] >= row["Coverage area proxy (km²)"]) &
                     (pdf["Average latency proxy (min)"] <= row["Average latency proxy (min)"]) &
                     ((pdf["Coverage area proxy (km²)"] > row["Coverage area proxy (km²)"]) |
                      (pdf["Average latency proxy (min)"] < row["Average latency proxy (min)"]))).any()
        pareto.append(not dominated)
    pdf["Pareto"] = pareto
    fig, ax = plt.subplots(figsize=(9, 5.5))
    ax.plot(pdf["Coverage area proxy (km²)"], pdf["Average latency proxy (min)"], marker="o", lw=1.8, label="Greedy allocation candidates")
    ax.scatter(pdf.loc[pdf["Pareto"], "Coverage area proxy (km²)"], pdf.loc[pdf["Pareto"], "Average latency proxy (min)"], s=90, marker="D", label="Pareto-efficient")
    for _, row in pdf.iterrows():
        ax.annotate(f"{int(row['Active patrol bases'])} bases", (row["Coverage area proxy (km²)"], row["Average latency proxy (min)"]), xytext=(4, 4), textcoords="offset points", fontsize=8)
    ax.set(title="RQ10 · Coverage–latency Pareto frontier (beat-centroid proxy)", xlabel="Geographic coverage-area proxy (km²; 5 km radius)", ylabel="Demand-weighted average latency proxy (minutes)")
    ax.legend()
    _save_fig(fig, outdir, "RQ10_Visualisation_1_Pareto_Coverage_vs_Latency.png")

    # Q-learning dispatch-zone policy: incident zone is the state; selected patrol zone is the action.
    n_zones = min(6, n)
    km = KMeans(n_clusters=n_zones, n_init=10, random_state=RANDOM_STATE)
    zone_id = km.fit_predict(beat_stats[["lat", "lon"]].to_numpy(), sample_weight=demand)
    zone_rows = []
    for z in range(n_zones):
        mask = zone_id == z
        w = demand[mask]
        zone_rows.append([np.average(beat_stats.loc[mask, "lat"], weights=w),
                          np.average(beat_stats.loc[mask, "lon"], weights=w), w.sum()])
    zdf = pd.DataFrame(zone_rows, columns=["lat", "lon", "demand"])
    zdist = haversine_distances(np.radians(zdf[["lat", "lon"]].to_numpy())) * 6371.0
    eta = zdist / 35.0 * 60.0 + 2.0
    pstate = zdf["demand"].to_numpy(dtype=float)
    pstate /= pstate.sum()
    Q = np.zeros((n_zones, n_zones), dtype=float)
    alpha, gamma = .12, .90
    rewards = np.zeros(10000, dtype=float)
    for t in range(10000):
        state = int(np.random.choice(n_zones, p=pstate))
        eps = max(.05, 1.0 - .95 * t / 7500)
        action = int(np.random.randint(n_zones)) if np.random.rand() < eps else int(np.argmax(Q[state]))
        reward = -float(eta[state, action])
        next_state = int(np.random.choice(n_zones, p=pstate))
        Q[state, action] += alpha * (reward + gamma * np.max(Q[next_state]) - Q[state, action])
        rewards[t] = reward
    cumulative_mean = np.cumsum(rewards) / np.arange(1, len(rewards)+1)
    rolling = pd.Series(rewards).rolling(250, min_periods=1).mean().to_numpy()
    fig, ax = plt.subplots(figsize=(10, 5))
    ax.plot(np.arange(1, 10001), cumulative_mean, lw=1.7, label="Cumulative mean reward")
    ax.plot(np.arange(1, 10001), rolling, lw=1.0, alpha=.75, label="250-step rolling mean")
    ax.set(title="RQ10 · Q-learning reward convergence (10,000 simulated dispatch steps)", xlabel="Simulation step", ylabel="Reward (negative estimated response minutes; higher is better)")
    ax.legend()
    _save_fig(fig, outdir, "RQ10_Visualisation_2_RL_Reward_Convergence.png")
    learned_actions = np.argmax(Q, axis=1)
    learned_eta = float(np.average(eta[np.arange(n_zones), learned_actions], weights=pstate))
    nearest_eta = float(np.average(eta.min(axis=1), weights=pstate))
    random_eta = float(np.average(eta.mean(axis=1), weights=pstate))
    first_mean = float(-rewards[:1000].mean())
    last_mean = float(-rewards[-1000:].mean())
    return f"Greedy allocation range: {pdf.iloc[0]['Average latency proxy (min)']:.2f} min with 1 base to {pdf.iloc[-1]['Average latency proxy (min)']:.2f} min with {max_k} bases; area coverage proxy {pdf.iloc[0]['Coverage area proxy (km²)']:.1f}–{pdf.iloc[-1]['Coverage area proxy (km²)']:.1f} km². Q-learning learned-policy estimated latency={learned_eta:.2f} min; nearest-zone theoretical lower bound={nearest_eta:.2f} min; random dispatch={random_eta:.2f} min. Simulation mean ETA moved from {first_mean:.2f} min in the first 1,000 steps to {last_mean:.2f} min in the last 1,000. Assumptions: 35 km/h straight-line speed, 2-minute overhead, 5 km coverage radius; these are not observed operational response times or official beat polygons."


def run_all_rqs(raw_df: pd.DataFrame, output_dir="rq_visualisations") -> pd.DataFrame:
    """Run all ten research question analyses and save exactly two figures per RQ."""
    outdir = Path(output_dir)
    outdir.mkdir(parents=True, exist_ok=True)
    d = _prepare_data(raw_df)
    print(f"Dataset rows: {len(d):,}; columns: {len(raw_df.columns)}")
    print(f"Valid start timestamps: {d['_start_dt'].notna().sum():,}")
    print(f"Valid Montgomery County coordinates: {_valid_coordinates(d).sum():,}")
    print(f"Valid engineered dispatch delays (0–1,440 min): {d['Dispatch_Delay_Minutes'].notna().sum():,}")
    print("\nIMPORTANT: results are data-derived only when this notebook is run against the CSV.\n"
          "RQ1 grid contours use coordinate bins smoothed as KDE-style density estimates.\n"
          "RQ3 graph adjacency is inferred from nearest Beat centroids, not official polygons.\n"
          "RQ10 uses a stated simulation proxy, not real dispatch vehicle travel logs.\n")

    results = {}
    tasks = [
        ("RQ1 — Spatial Forecast Accuracy", rq1_spatial_forecast),
        ("RQ2 — Temporal Discretization Variance", rq2_temporal_granularity),
        ("RQ3 — Network Graph Propagation", rq3_graph_propagation),
        ("RQ4 — Dispatch Urgency Conditioning", rq4_dispatch_urgency),
        ("RQ5 — Unsupervised Incident Isolation", rq5_anomaly_detection),
        ("RQ6 — Resampling Algorithm Generalization", rq6_resampling),
        ("RQ7 — Environmental Feature Attributions", rq7_place_attribution),
        ("RQ8 — Non-Stationary Data Drift", rq8_data_drift),
        ("RQ9 — SHAP Interpretability", rq9_shap_explainability),
        ("RQ10 — Reinforcement Routing Optimization", rq10_routing_optimization),
    ]
    for name, func in tasks:
        print("\n" + "#" * 88 + "\n" + name + "\n" + "#" * 88)
        try:
            results[name] = func(d, outdir)
            print("ANSWER:", results[name])
        except Exception as exc:
            results[name] = f"NOT COMPLETED: {type(exc).__name__}: {exc}"
            print("RQ ERROR:", results[name])
            import traceback
            traceback.print_exc(limit=2)

    summary = pd.DataFrame([{"Research Question": k, "Data-driven finding / status": v} for k, v in results.items()])
    summary_path = outdir / "RQ1_to_RQ10_Results_Summary.csv"
    summary.to_csv(summary_path, index=False)
    index_path = outdir / "visualisation_file_index.csv"
    index = pd.DataFrame([{"RQ": i, "Visualisation 1": f"RQ{i}_Visualisation_1_*.png", "Visualisation 2": f"RQ{i}_Visualisation_2_*.png"} for i in range(1, 11)])
    index.to_csv(index_path, index=False)
    print("\n" + "=" * 88 + "\nFINAL RESEARCH ANSWER SUMMARY\n" + "=" * 88)
    display(summary) if "display" in globals() else print(summary.to_string(index=False))
    figure_files = sorted(outdir.glob("RQ*_Visualisation_*.png"))
    zip_path = outdir / "All_20_RQ_Visualisations.zip"
    if figure_files:
        import zipfile
        with zipfile.ZipFile(zip_path, "w", compression=zipfile.ZIP_DEFLATED) as zf:
            for fp in figure_files:
                zf.write(fp, arcname=fp.name)
    print(f"\nFigure files created: {len(figure_files)} (target: 20)")
    print(f"Figure directory: {outdir.resolve()}")
    if figure_files:
        print(f"Visualisation ZIP: {zip_path.resolve()}")
    print(f"Summary CSV: {summary_path.resolve()}")
    return summary
```
